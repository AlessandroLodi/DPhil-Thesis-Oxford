```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1,1 , num=500, endpoint=True)
f = lambda x,a: 1 / (np.exp(x/a)+1)
temp_list = [0.01, 0.1, 0.2, 0.5, 1]

for T in temp_list:
    plt.plot(x, f(x,T),label=f'T={T}')
    plt.ylim(-0.1, 1.1)
plt.legend()
plt.suptitle('Fermi$-$Dirac Distribution')
```
